# Direct Corrections for Deadtime on Cloud Simulations

This code compares the Mueller and Correlated Deadtime Correction (CDC) that leverages the "active fraction histogram" from my paper.

I also want to look at how this compares to ABLE's specs. 100-120 MHz clock (so minimum 8.3 ns range resolution). Also, simulated fluxes off the surface are around 10 MHz highest albedo, 3-4 MHz for dark surfaces. A deadtime value of 20 ns is realistic. They want to push it down to 8-9 ns.

# Now calculate CDC values

```python
t_min = t_sim_bins[0]  # [s]
t_max = t_sim_bins[-1]  # [s]
window_bnd = [t_min, t_max]  # [s]

# calculate active fraction histogram
AF_range_res = 0.003  # [m]

AF_binwidth = 2 * AF_range_res / c  # [s]

intgrl_N = int(
    (t_max - t_min) / AF_binwidth
)

deadtime = tD  # [s] 

shot_avg = int(1470)
exclude_shots = True

# Check that shot average is divisible by the number of shots
if Nshot % shot_avg != 0:
    print('Change "shot_avg" variable to divisible number!')
    
if type(shot_avg) != int:
    shot_avg = int(shot_avg)
    
num_temporal_bins = Nshot / shot_avg
print('Number of temporal bins: {}'.format(num_temporal_bins))
print('Number of range bins: {}'.format(intgrl_N))
```

```python
avg_bins = 7500

if avg_bins > intgrl_N:
    raise ValueError(
        f'avg_bins ({avg_bins}) cannot exceed '
        f'intgrl_N ({intgrl_N}).'
    )

max_idx = (intgrl_N // avg_bins) * avg_bins

if max_idx == 0:
    raise ValueError(
        'No complete coarse range bins can be formed. '
        'Decrease avg_bins.'
    )

max_idx = (intgrl_N // avg_bins) * avg_bins
rmndr = intgrl_N - max_idx

print(
    'Remainder range bins per column getting thrown away '
    'to average to coarser bins: {}'.format(rmndr)
)

num_range_bins_coarse = intgrl_N // avg_bins

# Raw detected flux
cnts_curtain = np.zeros(
    (int(num_temporal_bins), num_range_bins_coarse)
)

# Mueller: average -> correct
muller_avg_correct_curtain = np.zeros(
    (int(num_temporal_bins), num_range_bins_coarse)
)

# Mueller: correct -> average
muller_correct_avg_curtain = np.zeros(
    (int(num_temporal_bins), num_range_bins_coarse)
)

# AFC: average -> correct
afc_avg_correct_curtain = np.zeros(
    (int(num_temporal_bins), num_range_bins_coarse)
)

# AFC: correct -> average
afc_correct_avg_curtain = np.zeros(
    (int(num_temporal_bins), num_range_bins_coarse)
)

# Store native-resolution active fraction if desired
active_ratio_curtain = np.zeros(
    (int(num_temporal_bins), intgrl_N)
)


for i in range(int(num_temporal_bins)):

    flight_time, n_shots, t_det_lst = dorg.data_organize(
        dt_sim,
        load_dir,
        fname,
        window_bnd,
        shot_avg - 1,
        exclude_shots,
        repeat_num=i + 1
    )

    t_phot_fit_tnsr = flight_time.values
    n_shots += 1

    # ---------------------------------------------------------
    # Native-resolution active fraction
    # ---------------------------------------------------------

    active_ratio_hst, AF_bin_edges = fit.deadtime_noise_hist(
        t_min,
        t_max,
        intgrl_N,
        deadtime,
        t_det_lst,
        n_shots
    )

    active_ratio_hst = active_ratio_hst.detach().numpy()

    # Prevent division by zero
    active_ratio_hst[
        np.where(active_ratio_hst == 0)[0]
    ] = sys.float_info.min

    active_ratio_curtain[i, :] = active_ratio_hst

    # ---------------------------------------------------------
    # Native-resolution detected counts
    # ---------------------------------------------------------

    Y = np.histogram(
        t_phot_fit_tnsr,
        bins=AF_bin_edges
    )[0]

    binwidth = np.diff(AF_bin_edges)[0]
    binwidth_smooth = binwidth * avg_bins

    # Native-resolution measured flux
    det_flux_fine = (
        Y
        / n_shots
        / binwidth
    )

    # =========================================================
    # 1. MUELLER: CORRECT -> AVERAGE
    # =========================================================

    muller_flux_fine = (
        det_flux_fine
        / (1 - det_flux_fine * deadtime)
    )

    muller_fine_reshape = np.reshape(
        muller_flux_fine[:max_idx],
        (max_idx // avg_bins, avg_bins)
    )

    # Average corrected fine-resolution flux
    muller_correct_avg = np.mean(
        muller_fine_reshape,
        axis=1
    )

    muller_correct_avg_curtain[i, :] = (
        muller_correct_avg
    )

    # =========================================================
    # 2. AFC: CORRECT -> AVERAGE
    # =========================================================

    afc_flux_fine = (
        det_flux_fine
        / active_ratio_hst
    )

    afc_fine_reshape = np.reshape(
        afc_flux_fine[:max_idx],
        (max_idx // avg_bins, avg_bins)
    )

    # Average corrected fine-resolution flux
    afc_correct_avg = np.mean(
        afc_fine_reshape,
        axis=1
    )

    afc_correct_avg_curtain[i, :] = (
        afc_correct_avg
    )

    # =========================================================
    # FIRST AVERAGE THE NATIVE-RESOLUTION DATA
    # =========================================================

    # Raw counts
    Y_reshape = np.reshape(
        Y[:max_idx],
        (max_idx // avg_bins, avg_bins)
    )

    Y_smooth = np.sum(
        Y_reshape,
        axis=1
    )

    # Coarse measured flux
    det_flux_coarse = (
        Y_smooth
        / n_shots
        / binwidth_smooth
    )

    cnts_curtain[i, :] = det_flux_coarse

    # ---------------------------------------------------------
    # Coarse active fraction
    # ---------------------------------------------------------

    active_ratio_reshape = np.reshape(
        active_ratio_hst[:max_idx],
        (max_idx // avg_bins, avg_bins)
    )

    active_ratio_coarse = np.mean(
        active_ratio_reshape,
        axis=1
    )

    # Prevent division by zero
    active_ratio_coarse[
        active_ratio_coarse == 0
    ] = sys.float_info.min

    # =========================================================
    # 3. MUELLER: AVERAGE -> CORRECT
    # =========================================================

    muller_avg_correct = (
        det_flux_coarse
        / (1 - det_flux_coarse * deadtime)
    )

    muller_avg_correct_curtain[i, :] = (
        muller_avg_correct
    )

    # =========================================================
    # 4. AFC: AVERAGE -> CORRECT
    # =========================================================

    afc_avg_correct = (
        det_flux_coarse
        / active_ratio_coarse
    )

    afc_avg_correct_curtain[i, :] = (
        afc_avg_correct
    )

    print('Time bin #{}'.format(i))


# -------------------------------------------------------------
# Orient curtains consistently
# -------------------------------------------------------------

cnts_curtain = cnts_curtain.T[::-1]

muller_avg_correct_curtain = (
    muller_avg_correct_curtain.T[::-1]
)

muller_correct_avg_curtain = (
    muller_correct_avg_curtain.T[::-1]
)

afc_avg_correct_curtain = (
    afc_avg_correct_curtain.T[::-1]
)

afc_correct_avg_curtain = (
    afc_correct_avg_curtain.T[::-1]
)
```

```python
# -------------------------------------------------------------
# Set up truth curtain at the coarse reporting resolution
# -------------------------------------------------------------

num_range_bins_coarse = intgrl_N // avg_bins

# Use the same truncated range as the processed data
truth_bin_edges = AF_bin_edges[:max_idx + 1:avg_bins]

curtain_truth = np.zeros((num_range_bins_coarse, Nshot))

for i in range(Nshot):

    n, _ = np.histogram(
        np.array(phot_events_ordered[i]) * dt_sim,
        bins=truth_bin_edges
    )

    curtain_truth[:, i] = n[::-1]


# -------------------------------------------------------------
# Average truth temporally
# -------------------------------------------------------------

rows, cols = curtain_truth.shape

curtain_truth = curtain_truth.reshape(
    rows,
    cols // shot_avg,
    shot_avg
).sum(axis=2)


# Convert truth counts to flux
curtain_truth_flux = (
    curtain_truth
    / shot_avg
    / binwidth_smooth
)


# -------------------------------------------------------------
# Range coordinate
# -------------------------------------------------------------

# Native correction-resolution binwidth
binwidth_native = np.diff(AF_bin_edges)[0]

# Coarse bin edges corresponding exactly to processing bins
truth_bin_edges = AF_bin_edges[:max_idx + 1:avg_bins]

# Coarse bin centers
t_plot = (
    truth_bin_edges[:-1]
    + np.diff(truth_bin_edges) / 2
)

range_plot = t_plot * c / 2  # [m]

# Print native and coarse range resolutions
native_range_res = np.diff(AF_bin_edges)[0] * c / 2
coarse_range_res = native_range_res * avg_bins

print('Native correction resolution: {:.4f} m'.format(native_range_res))
print('Coarse reporting resolution: {:.4f} m'.format(coarse_range_res))
print('Range averaging factor: {}'.format(avg_bins))

print('Temporal resolution: {:.4f} s'.format(shot_avg / f_L))

# -------------------------------------------------------------
# Plot limits
# -------------------------------------------------------------

flux_max = np.nanmax([
    np.nanmax(curtain_truth_flux),
    np.nanmax(cnts_curtain),
    np.nanmax(muller_avg_correct_curtain),
    np.nanmax(muller_correct_avg_curtain),
    np.nanmax(afc_avg_correct_curtain),
    np.nanmax(afc_correct_avg_curtain)
])
```

```python
Y_curtain = []
active_ratio_hst_curtain = []
for i in range(int(num_temporal_bins)):
    flight_time, n_shots, t_det_lst = dorg.data_organize(dt_sim, load_dir, fname, window_bnd, shot_avg-1, exclude_shots, repeat_num=i+1)
    t_phot_fit_tnsr = flight_time.values
    n_shots += 1

    active_ratio_hst, AF_bin_edges = fit.deadtime_noise_hist(t_min, t_max, intgrl_N, deadtime, t_det_lst, n_shots)
    active_ratio_hst = active_ratio_hst.detach().numpy()
    active_ratio_hst[np.where(active_ratio_hst==0)[0]] = sys.float_info.min  # if AF=0 for any bin, replace it with small positive value
    if i == 0:
        t_bins = AF_bin_edges[:-1] + 0.5*np.diff(AF_bin_edges)[0]
        binwidth = np.diff(AF_bin_edges)[0]
        t_intgrl = torch.tensor(t_bins)  # [s]
        
        print('Acquisition binwidth: {:.2E}'.format(binwidth))

    Y = np.histogram(t_phot_fit_tnsr, bins=AF_bin_edges)[0]
    
    Y_curtain.append(Y)
    active_ratio_hst_curtain.append(active_ratio_hst)
    
    print('Bin #{}'.format(i))
    
Y_curtain = np.array(Y_curtain).T[::-1]
active_ratio_hst_curtain = np.array(active_ratio_hst_curtain).T[::-1]
```

```python
# Let's apply the Mueller and Inst. Deadtime Corrections to the profiles

muller = (1/n_shots/binwidth) * Y_curtain/(1-Y_curtain/n_shots/binwidth*deadtime)  # [Hz]
dinst = (1/n_shots/binwidth) * Y_curtain/active_ratio_hst_curtain  # [Hz]
raw = (1/n_shots/binwidth) * Y_curtain  # [Hz]

dinst[np.where(dinst==np.inf)[0]] = np.nan  # if inf value, set to highest number instead

bins = AF_bin_edges
center = 0.5 * (bins[:-1] + bins[1:])
```

```python
avg_bins = 1000
binwidth_smooth = binwidth * avg_bins

max_idx = (len(raw)//avg_bins) * avg_bins
rmndr = len(raw) - max_idx  # remainder to discard (make sure this is a small number)
print('Remainder bins per column getting thrown away to average to coarser bins: {}'.format(rmndr))

row_smooth, col_smooth = raw[:max_idx].shape

# Raw counts
raw_reshape = np.reshape(raw[:max_idx], (row_smooth//avg_bins, col_smooth, avg_bins))
raw_smooth = np.sum(raw_reshape, axis=2) / avg_bins
# CDC counts
dinst_reshape = np.reshape(dinst[:max_idx], (row_smooth//avg_bins, col_smooth, avg_bins))
dinst_smooth = np.sum(dinst_reshape, axis=2) / avg_bins
# Mueller counts
muller_reshape = np.reshape(muller[:max_idx], (row_smooth//avg_bins, col_smooth, avg_bins))
muller_smooth = np.sum(muller_reshape, axis=2) / avg_bins
```
